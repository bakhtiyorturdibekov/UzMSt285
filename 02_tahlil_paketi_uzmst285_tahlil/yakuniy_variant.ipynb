```python
x = np.arange(256); Wsum = np.zeros((256, 256)); Dsum = np.zeros((256, 256)); lmax = []; dmax = []
for t in range(60 * HAJM):
    a = int(rng.integers(0, 128)) * 2 + 1; b = int(rng.integers(0, 128)) * 2 + 1; tt = int(rng.integers(0, 256))
    g = S[((-a * x * INV[(b + 0xC8 * x) % 256]) % 256) ^ tt]; W = lat(g, 8); D = ddt(g, 256)
    Wsum += W; Dsum += D; lmax.append(np.abs(W[1:]).max() / 256); dmax.append(D[1:].max() / 256)
print(f"Element g: belgilangan parametrlarda maks |c| {min(lmax):.3f}…{max(lmax):.3f}, maks p {min(dmax):.4f}…{max(dmax):.4f}")
print(f"Parametrlar bo‘yicha o‘rtacha: maks |c| = {np.abs(Wsum/len(lmax))[1:,1:].max()/256:.4f}, maks p = {(Dsum/len(dmax))[1:].max()/256:.4f}")
def bitsu(a, p):
    nb = 8 if p == 256 else 4; return ((a[:, :, None] >> np.arange(nb)) & 1).reshape(len(a), -1)
def keys(N, p):
    if p == 256: return kalitlar(rng, N, 256)
    k = rng.integers(0, 16, (N, 32)); k[:, 0] &= 0xE; k[:, 1] &= 0xE; return k
for p in (256, 16):
    c = Cfg(p, 'sbz'); N = 2000 * HAJM; k = keys(N, p); h = rng.integers(0, p, (N, 32)); bl = rng.integers(0, p, (N, 32)); tr = []
    iteration_y(h, bl, k.copy(), k, c, trace=tr, rounds=2); h0 = [t[2] for t in tr if t[0] == 'SBZ0'][0]; hr = [t[2] for t in tr if t[0] == 'Bog2'][0]
    bb = (np.abs((1 - 2 * bitsu(hr, p).astype(float)).T @ (1 - 2 * bitsu(h0, p).astype(float))) / N).max()
    h2 = h.copy(); h2[:, 0] ^= p // 2; t2 = []; iteration_y(h2, bl, k.copy(), k, c, trace=t2, rounds=2)
    akt = [(a[0], a[1], round(float(((a[2] ^ b[2]) != 0).sum(1).mean()), 1), round(float((((a[2] ^ b[2]) == 0) | ((a[2] ^ b[2]) == p // 2)).all(1).mean()), 3)) for a, b in zip(tr, t2)][:6]
    print(f"p={p}: 1 raunddan keyin eng yaxshi «bit → bit» |c| = {bb:.3f} (shovqin ≈{4.5/math.sqrt(N):.3f}); MSB ayirmasi (amal, raund, faol, P(MSB)): {akt}")
```
